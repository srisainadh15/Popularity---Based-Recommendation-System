# Popularity Based Recommendation System

-To implement the popularity based recommendation system using the dataset,the idea is to recommend the most popular based on number of ratings or the average rating score.Here how you can do this

-- Group the dataset by book-title and calculate the average rating and the number of ratings for each book
--sort the books based on either the average rating or the number of ratings
(whichever you defined as popularity).

--Recommend books that have highest popularity score


```python
book_ratings = df.groupby('Book-Title').agg({'Book-Rating':['mean','count']}).reset_index()
```

```python
book_ratings
```

```python
book_ratings.columns = ['Book-Title','Avg-Rating','Number']
```

```python
book_ratings_sorted = book_ratings.sort_values(by=['Avg-Rating','Number'],ascending=False)
```

```python
top_books = book_ratings_sorted.head(10)
```

```python
top_books_final = pd.merge(top_books,df[['Book-Title','Image-URL-M']],on='Book-Title',how='left')
```

```python
top_books_final
```

```python
final_books_recommend = top_books_final.drop_duplicates(['Book-Title'])
```

```python
final_books_recommend.to_csv('final_books_recommend.csv')
```
